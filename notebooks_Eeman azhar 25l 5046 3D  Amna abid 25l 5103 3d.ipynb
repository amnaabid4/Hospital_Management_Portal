# Hospital Management System

```python

# ---------- Data storage ----------
patients = []       
doctors = []        
appointments = []  

# Login details for the admin
ADMIN_USERNAME = "admin"
ADMIN_PASSWORD = "admin123"

# Charges used when generating a bill
CONSULTATION_FEE = 1500
ROOM_CHARGE_PER_DAY = 3000


# ---------- Login ----------

def login():
    
    print("===== Hospital Management System Login =====")
    for attempt in range(3):
        username = input("Username: ")
        password = input("Password: ")
        if username == ADMIN_USERNAME and password == ADMIN_PASSWORD:
            print("Login successful!\n")
            return True
        print("Wrong username or password. Attempts left:", 2 - attempt)
    return False


# ---------- Patient functions ----------

def add_patient():
    print("\n--- Add Patient ---")
    pid = input("Patient ID: ")

    # Check that the ID is not already used
    for p in patients:
        if p["id"] == pid:
            print("A patient with this ID already exists.\n")
            return

    name = input("Name: ")

    age = input("Age: ")
    while not age.isdigit():
        age = input("Please enter a number for age: ")
    age = int(age)

    gender = input("Gender (M/F): ")
    illness = input("Illness: ")

    patients.append({"id": pid, "name": name, "age": age,
                     "gender": gender, "illness": illness})
    print("Patient added successfully.\n")


def view_patients():
    print("\n--- All Patients ---")
    if len(patients) == 0:
        print("No patients registered yet.\n")
        return
    for p in patients:
        print("ID:", p["id"], "| Name:", p["name"], "| Age:", p["age"],
              "| Gender:", p["gender"], "| Illness:", p["illness"])
    print("Total patients:", len(patients), "\n")


def search_patient():
    print("\n--- Search Patient ---")
    print("1. Search by ID")
    print("2. Search by Name")
    choice = input("Enter choice: ")

    if choice == "1":
        pid = input("Enter patient ID: ")
        for p in patients:
            if p["id"] == pid:
                print("ID:", p["id"], "| Name:", p["name"], "| Age:", p["age"],
                      "| Gender:", p["gender"], "| Illness:", p["illness"], "\n")
                return
        print("Patient not found.\n")

    elif choice == "2":
        name = input("Enter name (or part of it): ").lower()
        found = False
        for p in patients:
            if name in p["name"].lower():
                print("ID:", p["id"], "| Name:", p["name"], "| Age:", p["age"],
                      "| Gender:", p["gender"], "| Illness:", p["illness"])
                found = True
        if not found:
            print("No patient found with that name.")
        print()

    else:
        print("Invalid choice.\n")


def update_patient():
    print("\n--- Update Patient ---")
    pid = input("Enter patient ID to update: ")

    for p in patients:
        if p["id"] == pid:
            print("Press Enter to keep the current value.")

            new_name = input("Name [" + p["name"] + "]: ")
            if new_name != "":
                p["name"] = new_name

            new_age = input("Age [" + str(p["age"]) + "]: ")
            if new_age != "":
                if new_age.isdigit():
                    p["age"] = int(new_age)
                else:
                    print("Invalid age, keeping the old value.")

            new_gender = input("Gender [" + p["gender"] + "]: ")
            if new_gender != "":
                p["gender"] = new_gender

            new_illness = input("Illness [" + p["illness"] + "]: ")
            if new_illness != "":
                p["illness"] = new_illness

            print("Patient updated successfully.\n")
            return

    print("Patient not found.\n")


def delete_patient():
    print("\n--- Discharge / Delete Patient ---")
    pid = input("Enter patient ID to delete: ")

    for i in range(len(patients)):
        if patients[i]["id"] == pid:
            confirm = input("Are you sure you want to delete " + patients[i]["name"] + "? (y/n): ")
            if confirm.lower() != "y":
                print("Delete cancelled.\n")
                return

            patients.pop(i)

            # Also remove this patient's appointments (loop backwards so positions don't shift)
            j = len(appointments) - 1
            while j >= 0:
                if appointments[j]["patient_id"] == pid:
                    appointments.pop(j)
                j = j - 1

            print("Patient deleted successfully.\n")
            return

    print("Patient not found.\n")


# ---------- Doctor functions ----------

def add_doctor():
    print("\n--- Add Doctor ---")
    did = input("Doctor ID: ")

    # Check that the ID is not already used
    for d in doctors:
        if d["id"] == did:
            print("A doctor with this ID already exists.\n")
            return

    name = input("Name: ")
    specialization = input("Specialization: ")
    doctors.append({"id": did, "name": name, "specialization": specialization})
    print("Doctor added successfully.\n")


def view_doctors():
    print("\n--- All Doctors ---")
    if len(doctors) == 0:
        print("No doctors registered yet.\n")
        return
    for d in doctors:
        print("ID:", d["id"], "| Name: Dr.", d["name"], "| Specialization:", d["specialization"])
    print("Total doctors:", len(doctors), "\n")


# ---------- Appointment functions ----------

def book_appointment():
    print("\n--- Book Appointment ---")
    if len(patients) == 0 or len(doctors) == 0:
        print("You need at least one patient and one doctor first.\n")
        return

    # Find the patient
    pid = input("Patient ID: ")
    patient_name = ""
    for p in patients:
        if p["id"] == pid:
            patient_name = p["name"]
    if patient_name == "":
        print("Patient not found.\n")
        return

    # Find the doctor
    did = input("Doctor ID: ")
    doctor_name = ""
    for d in doctors:
        if d["id"] == did:
            doctor_name = d["name"]
    if doctor_name == "":
        print("Doctor not found.\n")
        return

    date = input("Date (DD-MM-YYYY): ")
    time = input("Time (e.g. 10:30 AM): ")

    # A doctor cannot have two appointments at the same date and time
    for a in appointments:
        if a["doctor_id"] == did and a["date"] == date and a["time"] == time:
            print("This doctor is already booked at that time.\n")
            return

    appointments.append({"patient_id": pid, "patient_name": patient_name,
                         "doctor_id": did, "doctor_name": doctor_name,
                         "date": date, "time": time})
    print("Appointment booked successfully.\n")


def view_appointments():
    print("\n--- All Appointments ---")
    if len(appointments) == 0:
        print("No appointments booked yet.\n")
        return
    number = 1
    for a in appointments:
        print(str(number) + ".", a["patient_name"], "with Dr.", a["doctor_name"],
              "on", a["date"], "at", a["time"])
        number = number + 1
    print()


def cancel_appointment():
    print("\n--- Cancel Appointment ---")
    if len(appointments) == 0:
        print("No appointments to cancel.\n")
        return

    view_appointments()

    number = input("Enter appointment number to cancel: ")
    while not number.isdigit():
        number = input("Please enter a number: ")
    number = int(number)

    if number < 1 or number > len(appointments):
        print("Invalid appointment number.\n")
        return
    appointments.pop(number - 1)
    print("Appointment cancelled.\n")


# ---------- Billing ----------

def generate_bill():
    print("\n--- Generate Bill ---")
    pid = input("Enter patient ID: ")

    for p in patients:
        if p["id"] == pid:
            days = input("Number of days admitted (0 if not admitted): ")
            while not days.isdigit():
                days = input("Please enter a number of days: ")
            days = int(days)

            medicine = input("Medicine charges: ")
            while not medicine.isdigit():
                medicine = input("Please enter a number for medicine charges: ")
            medicine = int(medicine)

            room_total = days * ROOM_CHARGE_PER_DAY
            total = CONSULTATION_FEE + room_total + medicine

            print("\n========== HOSPITAL BILL ==========")
            print("Patient:", p["name"], "(ID:", pid + ")")
            print("Consultation fee:   ", CONSULTATION_FEE)
            print("Room charges:       ", room_total, "(" + str(days), "days x", str(ROOM_CHARGE_PER_DAY) + ")")
            print("Medicine charges:   ", medicine)
            print("-----------------------------------")
            print("TOTAL:              ", total)
            print("===================================\n")
            return

    print("Patient not found.\n")


# ---------- Main menu ----------

def main():
    if not login():
        print("Too many failed attempts. Exiting program.")
        return

    while True:
        print("========== Hospital Management System ==========")
        print("1.  Add Patient")
        print("2.  View All Patients")
        print("3.  Search Patient")
        print("4.  Update Patient")
        print("5.  Discharge / Delete Patient")
        print("6.  Add Doctor")
        print("7.  View All Doctors")
        print("8.  Book Appointment")
        print("9.  View Appointments")
        print("10. Cancel Appointment")
        print("11. Generate Bill")
        print("0.  Exit")
        choice = input("Enter your choice: ")

        if choice == "1":
            add_patient()
        elif choice == "2":
            view_patients()
        elif choice == "3":
            search_patient()
        elif choice == "4":
            update_patient()
        elif choice == "5":
            delete_patient()
        elif choice == "6":
            add_doctor()
        elif choice == "7":
            view_doctors()
        elif choice == "8":
            book_appointment()
        elif choice == "9":
            view_appointments()
        elif choice == "10":
            cancel_appointment()
        elif choice == "11":
            generate_bill()
        elif choice == "0":
            print("Thank you for using the Hospital Management System. Goodbye!")
            break
        else:
            print("Invalid choice, please try again.\n")


main()
```
